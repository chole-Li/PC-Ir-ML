```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd
import os

# Read input data from an Excel file
input_data = pd.read_excel('smiles.xlsx')


# Directory to save the output CSV files
output_directory = r"."  # Replace with your desired directory

for index, row in input_data.iterrows():
    smiles_string = row['SMILES']
    output_file_name = row['OutputFile']

    mol = Chem.MolFromSmiles(smiles_string)

    if mol is not None:
        # Calculate all available descriptors using Descriptors module
        all_descriptors = {}
        for descriptor, function in Descriptors.descList:
            try:
                all_descriptors[descriptor] = function(mol)
            except:
                all_descriptors[descriptor] = None
        
        df = pd.DataFrame(all_descriptors, index=[0])

        output_file_path = os.path.join(output_directory, f'{output_file_name}.csv')
        df.to_csv(output_file_path, index=False)

print("All molecule descriptors saved.")
```
